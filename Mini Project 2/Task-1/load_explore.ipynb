# Task 1 — Load and Explore Dataset (ChurnGuard)

## Dataset Overview
- **File:** churnguard_data.csv
- **Shape:** 1030 rows, 12 columns
- **Goal:** Identify customers likely to cancel subscription (Churn prediction)

## Column Summary
| Column | Type | Description |
|--------|------|-------------|
| customerID | object | Unique customer ID |
| gender | object | Customer gender |
| SeniorCitizen | int64 | 1=senior, 0=not |
| tenure | float64 | Months with company |
| PhoneService | object | Has phone service |
| InternetService | object | Type of internet |
| Contract | object | Contract type |
| PaperlessBilling | object | Paperless billing |
| PaymentMethod | object | Payment method |
| MonthlyCharges | float64 | Monthly bill |
| TotalCharges | object | Total billed (stored as TEXT!) |
| Churn | object | Target — left or not |

## Data Quality Issues Found

### 1. Missing Values
| Column | Missing Count |
|--------|--------------|
| tenure | 50 |
| MonthlyCharges | 75 |
| TotalCharges | 64 |
| InternetService | 16 |

### 2. Duplicate Rows
30 duplicate rows found → need to remove!
### 3. Churn Column — Inconsistent Casing
No, NO, no, nO   → all mean "No"
Yes, YES, yes, yEs → all mean "Yes"
Should be only: Yes / No

### 4. Contract Column — Typos and Inconsistent Formats
12 unique values found → should be only 3!
Month-to-month = month to month = month-to-month = Monthly
One year       = One Year       = one year       = 1 year
Two year       = Two Year       = two year       = 2 year

### 5. tenure — Impossible Values
min = -5 (negative tenure is impossible!)
→ needs to be handled

### 6. MonthlyCharges — Possible Outlier
max = 1500 (very high compared to mean=70)
→ needs investigation

### 7. TotalCharges — Wrong Data Type
Stored as object (text) instead of float
→ needs conversion to numeric
